# **1. 프롬프트 엔지니어링**
- 인공지능 모델, 특히 자연어 처리 모델과의 상호작용을 최적화하는 기술
- 모델이 의도한 답변을 도축하도록 텍스트를 구성하는 방식으로, 적절한 질문 형식이나 설명을 제공하여 더 나은 성과를 얻는 것이 목표로 함

### 1. Role
AI모델과의 대화에서 특정 문맥을 설정하거나, 모델의 응답 스타일과 관점을 조정하기 위해 사용하는 개념
- User: AI와 상호작용하는 주체로, 일반적인 질문이나 요청을 전달
- Assistant: AI 모델이 자체적으로 응답할 때의 역활로, 사용자의 질문이나 명령에 따라 답변을 제공
- System: AI 모델의 행동과 스타일을 지시하는 역할로, AI가 어떻게 응답해야 할지 지침을 설정.x

### 토큰
- 토큰은 모델이 처리하는 텍스트의 작은 단위
- 하나의 토큰은 단어, 부분적인 단어, 또는 공백이나 구두점일 수 있음
- GPT 모델은 텍스트를 처리할 때 토큰들을 기반으로 학습하고 예측 생성

````python
from dotenv import load_dotenv
import os
from openai import OpenAI
````

````python
load_dotenv()
api_key = os.environ["OPENAI_API_KEY"]
````

````python
client = OpenAI(
    api_key=os.environ["OPENAI_API_KEY"]
)

"""
제 이름은 송정근입니다.
안녕하세요. 송정근님
제 이름이 뭐였죠?
"""

messages = [
    {
        "role": "system",
        "content": "당신은 초등학교 선생님입니다."
    },
    {
        "role": "user",
        "content": "FastAPI가 뭐예요?"
    }
]

response = client.responses.create(
    model="gpt-5-nano",
    # role:user 는 default 여서 안적어도 됨
    # messages=[{"role":"user", "content":"안녕하세요? 오늘 기분이 어때요?"}]
    input=messages
)

response
````

````python
response.output[1].content[0].text
````

````python
def get_response_text(response):
    for item in response.output:
        if item.type == "message":
            return item.content[0].text
    return None
        
get_response_text(response)
````

````python
# max_output_tokens: 토큰 제한

client = OpenAI(
    api_key=os.environ["OPENAI_API_KEY"]
)

messages = [
    {
        "role": "user",
        "content": "안녕하세요"
    }
]

response = client.responses.create(
    model="gpt-5-nano",
    # role:user 는 default 여서 안적어도 됨
    messages=[
        {
            "role": "user",
            "content": "안녕하세요"
        }
    ],
    max_output_tokens=30
)

# response
get_response_text(response)
````

````python
response.status
````

### response.status
- completed: 응답 생성이 정삭적으로 완료됨
- incomplete: 응답을 끝가지 만들지 못하고 중간에 종료됨
- failed: 오류로 인해 응답 생성 실패
- queued: 처리 대기열에 들어가 기다리는 중
- in_progress: 현재 응답을 생성하는 중
- cancelled: 요청이 취소됨

````python
# temperature: 항상 비슷한 답, 불필요하게 길게 말하는 경우 감소

client = OpenAI(
    api_key=os.environ["OPENAI_API_KEY"]
)

messages = [
    {
        "role": "user",
        "content": "안녕하세요"
    }
]

response = client.responses.create(
    model="gpt-4.1-nano",
    # role:user 는 default 여서 안적어도 됨
    input=messages,
    temperature=0
)

# response
get_response_text(response)
````

````python
with gr.Blocks() as demo:
    gr.Markdown("# 안녕하세요 Gradio!")
    gr.Markdown("### 작은 제목을 입력합니다")
    gr.Markdown("- 첫 번째 아이템\n - 두 번째 아이템\n - 세 번째 아이템")
    
demo.launch() 
````

````python
def handle_input(text):
    return text

with gr.Blocks() as demo:
    text_input = gr.Textbox(label="문자입력", lines=1)
    output_text = gr.Textbox(label="출력")
    text_input.submit(handle_input, inputs=text_input, outputs=output_text)
    
demo.launch() 
````

````python
def handle_checkbox(selected):
    if selected:
        return "동의했습니다."
    return "동의하지 않았습니다."

with gr.Blocks() as demo:
    checkbox= gr.Checkbox(label="개인정보 사용에 동의하겠습니까?")
    output_checkbox = gr.Textbox(label="출력")
    checkbox.change(handle_checkbox, inputs=checkbox, outputs=output_checkbox)

demo.launch()
````

````python
def handle_fruit(fruit):
    return f'선택한 과일: {fruit}'

with gr.Blocks() as demo:
    fruit_dropdown = gr.Dropdown(label="과일", choices=['사과', '오렌지', '바나나', '메론'])
    output_fruit = gr.Textbox(label='구입한 과일')
    fruit_dropdown.change(handle_fruit, inputs=fruit_dropdown, outputs=output_fruit)
    
demo.launch()
````

# **3.번역 앱 예제**

````python
# Few-shot 번역 예제 데이더
example = {
    "한국어": [
        "오늘 날씨 어때?",
        "최근 멀티모달과 Agent 기술이 인기를 끌고 있다."
    ],
    "영어": [
        "How is the weather today?",
        "Recently, multimodal and agent technologies have become popular."
    ],
    "일본어": [
        "今日の天気はどう？",
        "最近、マルチモーダルとAgent技術が人気を集めている."
    ]
}


def build_fewshot(src_lang, trg_lang):
    src_examples = example[src_lang]
    trg_examples = example[trg_lang]
    fewshot_message = []

    for src_text, trg_text in zip(src_examples, trg_examples):
        fewshot_message.append({
            "role": "user",
            "content": src_text
        })

        fewshot_message.append({
            "role": "assistant",
            "content": trg_text
        })

        return fewshot_message


def translate_text_chatgpt(text, src_lang, trg_lang):
    system_instruction = (
        "당신은 전문 번역 애플리케이션입니다,\n"
        f"입력된 {src_lang} 문장을 자연스러운 {trg_lang} 문장으로 번역하세요.\n"
        "설명, 부연 설명, 따옴표, 제목을 추가하지 마세요.\n"
        "번역된 텍스트만 출력하세요."
    )

    fewshot_messages = build_fewshot(
        src_lang=src_lang, trg_lang=trg_lang
    )

    messages = [
        {"role": "developer", "content": system_instruction},
        *fewshot_messages,
        {"role": "user", "content": text}
    ]

    try:
        response = client.responses.create(
            model="gpt-5-nano",
            input=messages,
            reasoning={"effort": "low"},
            max_output_tokens=500
        )

        translate_text = response.output_text.strip()
        if not translate_text:
            return "번역 결과가 생성되지 않았습니다. 다시 시도해 주세요."

        return translate_text

    except Exception as e:
        print(f'OpenAI API 오류: {e}')
        return f'번역 중 오류가 발생했습니다.: {e}'


def gradio_translate(text, src_lang, trg_lang):
    if text is None or not text.strip():
        return "번역할 내용일 입력하세요."

    if src_lang == trg_lang:
        return "원본 언어와 번역 언어가 같습니다. 다른 언어를 선택하세요."

    return translate_text_chatgpt(text=text.strip(), src_lang=src_lang, trg_lang=trg_lang)


with gr.Blocks() as demo:
    gr.Markdown("# 초간단 번역 앱")
    text_input = gr.Textbox(label="번역할 내용을 입력하세요",
                            placeholder="여기에 내용을 입력하세요", lines=3)
    src_lang_dropdown = gr.Dropdown(
        choices=["한국어", "영어", "일본어"], value="영어", label="원본 언어를 선택하세요")
    trg_lang_dropdown = gr.Dropdown(
        choices=["한국어", "영어", "일본어"], value="한국어", label="번역할 언어를 선택하세요")
    translate_button = gr.Button("번역하기")
    output_text = gr.Textbox(
        label="번역 결과", placeholder="번역된 결과가 여기에 표시됩니다", lines=5, interactive=False)
    translate_button.click(
        fn=gradio_translate, inputs=[
            text_input, src_lang_dropdown, trg_lang_dropdown], outputs=output_text
    )
    
demo.launch()
````

### GPT API 기반 AI 광고 문구 생성 서비스 구현

1. 프로젝트 구성
- Gradio 클라이언트
- FastAPI 서버

2. Gradio 화면 구현
- 제품 이름 입력창
- 제품 주요 내용 입력창
- 광고 문구의 느낌을 선택하는 체크박스 그룹(재밌게, 과장스럽게, 참신하게, 고급스럽게, 센스있게, 신선하게, 전문성있게...)
- 광고 문구 생성 버튼
- 생성된 광고 문구 출력창
- MongoDB에 저장된 최근 광고 문구 목록

3. FastAPI 요청 모델 구현
- product_name      (제품 이름)
- details           (제품 주요 내용)
- tone_and_manner   (체크 박스에 대한 데이터)

4. GPT API 광고 문구 생성
- 제품이름, 제품 주요 내용, 광고 문구 스타일의 정보가 모델에 전달되어야 함

5. MongoDB 저장
- 생성된 광고 문구와 입력 데이터를 MongoDB에 저장
- 제품 이름, 제품 주요 내용, 광고 문구의 느낌, 생성된 광고 문구 저장

6. 최신 데이터 조회
